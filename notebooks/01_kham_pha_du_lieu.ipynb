```python
import sys
from pathlib import Path
import pandas as pd

root = Path.cwd()
if root.name == 'notebooks':
    root = root.parent
sys.path.insert(0, str(root))

from src.prepare_data import GENRES, load_movielens
from src.features import temporal_split

ratings, users, movies = load_movielens()
pd.Series({'Lượt chấm': len(ratings), 'Người dùng': len(users), 'Phim': len(movies),
           'Cặp người và phim bị trùng': ratings.duplicated(['user_id', 'movie_id']).sum(),
           'Phim thiếu ngày phát hành': movies['release_date'].isna().sum()})
```

```python
ratings['rating'].value_counts().sort_index().plot.bar(
    xlabel='Số sao', ylabel='Số lượt chấm', title='Phân bố điểm MovieLens 100K');
```

```python
train, validation, warm, cold, *_ = temporal_split(ratings)
pd.DataFrame([{'Tập': name, 'Số lượt chấm': len(frame),
               'Số người dùng': frame['user_id'].nunique(),
               'Tỷ lệ thích': frame['liked'].mean()}
              for name, frame in [('train', train), ('validation', validation),
                                  ('warm test', warm), ('cold test', cold)]])
```

```python
movies[list(GENRES)].sum().sort_values().plot.barh(
    figsize=(7, 5), xlabel='Số phim', title='Phân bố thể loại phim');
```
